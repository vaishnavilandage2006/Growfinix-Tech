```python
import pandas as pd

df = pd.read_csv("students.csv")

df["Attendance"] = "Absent"

name = input("Enter student name: ").strip().lower()

for i in range(len(df)):
    if df.loc[i, "Name"].lower() == name:
        df.loc[i, "Attendance"] = "Present"

df.to_csv("updated_attendance.csv", index=False)

print(df)
```
