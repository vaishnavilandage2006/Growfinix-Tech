## Problem Statement

Many students face academic difficulties.

Teachers often identify problems late.

An AI system can analyze student data and predict
which students need additional support.

Data Collection

The AI system collects:

- Attendance
- Assignment Scores
- Internal Marks
- Study Hours
- Class Participation

```python
import pandas as pd

data = {
    "Attendance":[95,80,60,90,50],
    "Study_Hours":[5,4,2,6,1],
    "Marks":[88,75,45,92,35]
}

df = pd.DataFrame(data)

print(df)
```

```python
for mark in df["Marks"]:
    if mark >= 50:
        print("Pass")
    else:
        print("Needs Improvement")
```
